```python
# DMV EXP 5 - AQI Trends Visualization - City_Air_Quality.csv
import pandas as pd
import matplotlib.pyplot as plt

# 1. Import
df = pd.read_csv('City_Air_Quality.csv')

# 2. Explore
print("Shape:", df.shape)
print(df.head())
print(df.info())

# 3. Identify variables
df['Date'] = pd.to_datetime(df['Date'])

# 4. Line plot - Overall AQI trend over time
plt.figure(figsize=(8,4))
plt.plot(df['Date'], df['AQI'], marker='o', color='red')
plt.title('Overall AQI Trend Over Time')
plt.xlabel('Date')
plt.ylabel('AQI')
plt.xticks(rotation=45)
plt.grid(True)
plt.show()

# 5. Individual pollutant trends
plt.figure(figsize=(8,4))
plt.plot(df['Date'], df['PM2.5'], label='PM2.5', color='blue')
plt.plot(df['Date'], df['PM10'], label='PM10', color='green')
plt.plot(df['Date'], df['CO'], label='CO', color='orange')
plt.title('Pollutant Levels Over Time')
plt.xlabel('Date')
plt.ylabel('Pollutant Levels')
plt.legend()
plt.xticks(rotation=45)
plt.show()

# 6. Bar plot - AQI across dates (first 10 days)
plt.figure(figsize=(8,4))
plt.bar(df['Date'].head(10).astype(str), df['AQI'].head(10), color='purple')
plt.title('AQI Comparison - First 10 Days')
plt.xlabel('Date')
plt.ylabel('AQI')
plt.xticks(rotation=45)
plt.show()

# 7. Box plot - Distribution of AQI
plt.figure(figsize=(5,4))
plt.boxplot(df['AQI'], patch_artist=True, boxprops=dict(facecolor='lightgreen'))
plt.title('Distribution of AQI Values')
plt.ylabel('AQI')
plt.show()

# 8. Scatter plot - AQI vs Pollutant
plt.figure(figsize=(6,4))
plt.scatter(df['PM2.5'], df['AQI'], color='red', alpha=0.6)
plt.title('AQI vs PM2.5')
plt.xlabel('PM2.5')
plt.ylabel('AQI')
plt.show()

plt.figure(figsize=(6,4))
plt.scatter(df['PM10'], df['AQI'], color='blue', alpha=0.6)
plt.title('AQI vs PM10')
plt.xlabel('PM10')
plt.ylabel('AQI')
plt.show()

print("All Visualizations Done")
```
