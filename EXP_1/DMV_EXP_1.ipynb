```python
# DMV EXP 1 - EASIEST VERSION - 1 Cell
import pandas as pd
import matplotlib.pyplot as plt

# 1. Create Data
data = {
'Order_ID': [101,102,103,104,105,106,107,108,109,110],
'Product_Category': ['Electronics','Clothing','Grocery','Furniture','Books','Sports','Electronics','Clothing','Grocery','Furniture'],
'Quantity': [2,3,5,1,4,2,1,5,3,2],
'Price': [50000,1500,200,12000,500,3000,25000,2000,300,15000]
}
df = pd.DataFrame(data)

# 2. Save to 3 formats
df.to_csv('Sales_Data.csv', index=False)
df.to_excel('Sales_Data.xlsx', index=False)
df.to_json('Sales_Data.json', orient='records')
print("Files Created")

# 3. Load all 3 formats
csv_data = pd.read_csv('Sales_Data.csv')
excel_data = pd.read_excel('Sales_Data.xlsx')
json_data = pd.read_json('Sales_Data.json')
print("CSV, Excel, JSON Loaded")

# 4. Cleaning & Transformation
df = df.drop_duplicates()
df['Total_Sales'] = df['Quantity'] * df['Price']
print("\n", df.head())

# 5. Analysis
print("\nTotal Sales:", df['Total_Sales'].sum())
print("Avg Order:", df['Total_Sales'].mean())
print(df['Product_Category'].value_counts())

# 6. Visualization - FIXED No Overlap
# Bar Plot
plt.figure(figsize=(6,4))
df.groupby('Product_Category')['Total_Sales'].sum().plot(kind='bar', color='skyblue')
plt.title('Sales by Product Category')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# Pie Chart - Separate Big Figure
plt.figure(figsize=(5,5))
counts = df['Product_Category'].value_counts()
plt.pie(counts, labels=counts.index, autopct='%1.1f%%', startangle=90)
plt.title('Product Category Distribution')
plt.show()

# Box Plot
plt.figure(figsize=(4,4))
plt.boxplot(df['Total_Sales'], patch_artist=True)
plt.title('Box Plot of Total Sales')
plt.ylabel('Total Sales')
plt.show()
```
